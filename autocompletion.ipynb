```python
text_data = "hello hellooo hellna hooy"
```

```python
#function to extract words from the text
def extract_words(text):
  words = []
  word=""
  for char in text:
    if char.isalpha():
      word += char
    else:
      if word:
        words.append(word)
        word = ""
  if word:
    words.append(word)
  return words

#stores distinct words in the text
distinct_words = set(extract_words(text_data))

```

```python
#uses O(n*len(word)) time complexity
#n-no of words,compares everyword in the vocabulary
#function returns possible completion of current word
def word_completion(curr_word,words):
  outs = []
  for word in words:
    if len(curr_word)>len(word):
      continue
    if word[:len(curr_word)]==curr_word:
      outs.append(word)
  return outs

curr_word = input("enter word to complete:")
for completion in word_completion(curr_word,distinct_words):
  print(completion)
```

```python
#defining TrieNode
class TrieNode:
  def __init__(self):
    self.children = {}
    self.isEnd = False
```

```python
#construct Trie of words
def constructTrie(words):
  root = TrieNode()
  for word in words:
    curr = root
    for ch in word:
      if ch not in curr.children:
        curr.children[ch] = TrieNode()
      curr = curr.children[ch]
    curr.isEnd = True
  return root
```

```python
#function to check whether the word exist in vocabulary
def wordExist(root,word):
  for ch in word:
    if ch not in root.children:
      return False
    root = root.children[ch]
  if root.isEnd:
    return True
  return False
```

```python
root = constructTrie(distinct_words)
word = input("enter word to check:")
if wordExist(root,word):
  print(f"{word} exists in vocabulary")
else:
  print(f"{word} does not exist in vocabulary")
```

```python
#construct words by bfs traversal
def dfs(root,word,words):
  if root.isEnd:
    words.append(word)
  for ch in root.children:
    dfs(root.children[ch],word+ch,words)

#function to autocomplete words
def autoCompleteWord(root,word):
  curr = root
  for ch in word:
    if ch not in curr.children:
      return [word]
    curr = curr.children[ch]
  completion = []
  dfs(curr,word,completion)
  return completion

```

```python
from collections import deque
class AutoCompletion:

  class TrieNode:
    def __init__(self):
      self.children = {}
      self.end = False

  def __init__(self,text,n):
    self.n = n
    self.text = text
    self.distinct_words = self.extract_words()
    self.root = self.construct_trie()
    self.sentences = self.extract_sentences()
    self.ngram = self.ngram_model()

  def extract_words(self):
    words = []
    word = ""
    for ch in self.text:
      if ch.isalpha():
        word += ch
      else:
        if word:
          words.append(word)
          word = ""
    if word:
      words.append(word)
    return set(words)

  def construct_trie(self):
    root = AutoCompletion.TrieNode()
    for word in self.distinct_words:
      curr = root
      for ch in word:
        if ch not in curr.children:
          curr.children[ch] = AutoCompletion.TrieNode()
        curr = curr.children[ch]
      curr.end = True
    return root

  def dfs(self,root,word,words):
    if root.end:
      words.append(word)

    for ch in root.children:
      self.dfs(root.children[ch],word+ch,words)



  def word_completion(self,curr_word):
    root = self.root
    for ch in curr_word:
      if ch not in root.children:
        return [word]
      root = root.children[ch]
    outs = []
    self.dfs(root,curr_word,outs)
    return outs

  def construct_key(self,words):
    key = ""
    for word in words:
      key += word
      key += " "
    return key
  def ngram_model(self):
    ngrams = {}
    for sentence in self.sentences:
      words = sentence.split()
      q = deque()
      if len(words) < self.n+1:
        continue
      for i in range(self.n):
        q.append(words[i])

      for i in range(self.n,len(words)):
        key = self.construct_key(q)
        if key not in ngrams:
          ngrams[key] = {}
        if words[i] not in ngrams[key]:
          ngrams[key][words[i]] = 1
        else:
          ngrams[key][words[i]] += 1
        q.popleft()
        q.append(words[i])
    for key in ngrams:
      ngrams[key] = dict(
          sorted(
              ngrams[key].items(),
              key=lambda item: item[1],
              reverse=True
          )
      )
    return ngrams


  def extract_sentences(self):
    sentences = []
    sentence = ""
    for ch in self.text:
      if ch.isalpha() or ch == " ":
        sentence += ch
      else:
        if sentence:
          sentences.append(sentence)
          sentence = ""
    if sentence:
      sentences.append(sentence)
    return set(sentences)

  def next_word_options(self,sentence):
    words = sentence.split()
    if len(words) < self.n:
      return []
    key = self.construct_key(words[-self.n:]);
    if key not in self.ngram:
      return []
    next_words = []
    for k in self.ngram[key]:
      next_words.append(k)
    return next_words

  def give_suggestions(self,sentence):
    if sentence[len(sentence)-1].isalpha():
      return self.word_completion(sentence)
    elif sentence[len(sentence)-1] == " ":
      return self.next_word_options(sentence)
    else:
      return []



```

```python
text = "hello good morning.how are you?i am doing good.i ran 4km today.good afternoon."
n = 1
obj = AutoCompletion(text,n)
inp = input("enter text:")
print(obj.give_suggestions(inp))
```
