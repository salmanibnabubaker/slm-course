```python
text_data = "hello hellooo hellna hooy"
```

```python
#function to extract words from the text
def extract_words(text):
  words = []
  word=""
  for char in text:
    if char.isalpha():
      word += char
    else:
      if word:
        words.append(word)
        word = ""
  if word:
    words.append(word)
  return words

#stores distinct words in the text
distinct_words = set(extract_words(text_data))

```

```python
#uses O(n*len(word)) time complexity
#n-no of words,compares everyword in the vocabulary
#function returns possible completion of current word
def word_completion(curr_word,words):
  outs = []
  for word in words:
    if len(curr_word)>len(word):
      continue
    if word[:len(curr_word)]==curr_word:
      outs.append(word)
  return outs

curr_word = input("enter word to complete:")
for completion in word_completion(curr_word,distinct_words):
  print(completion)
```

```python
#defining TrieNode
class TrieNode:
  def __init__(self):
    self.children = {}
    self.isEnd = False
```

```python
#construct Trie of words
def constructTrie(words):
  root = TrieNode()
  for word in words:
    curr = root
    for ch in word:
      if ch not in curr.children:
        curr.children[ch] = TrieNode()
      curr = curr.children[ch]
    curr.isEnd = True
  return root
```

```python
#function to check whether the word exist in vocabulary
def wordExist(root,word):
  for ch in word:
    if ch not in root.children:
      return False
    root = root.children[ch]
  if root.isEnd:
    return True
  return False
```

```python
root = constructTrie(distinct_words)
word = input("enter word to check:")
if wordExist(root,word):
  print(f"{word} exists in vocabulary")
else:
  print(f"{word} does not exist in vocabulary")
```

```python
#construct words by bfs traversal
def dfs(root,word,words):
  if root.isEnd:
    words.append(word)
  for ch in root.children:
    dfs(root.children[ch],word+ch,words)

#function to autocomplete words
def autoCompleteWord(root,word):
  curr = root
  for ch in word:
    if ch not in curr.children:
      return [word]
    curr = curr.children[ch]
  completion = []
  dfs(curr,word,completion)
  return completion

```

```python
inp = input("enter word to complete:")
completions = autoCompleteWord(root,inp)
print(completions)
```
