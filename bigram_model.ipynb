```python
#entire textual data
text_data = "hello my name is salman my name is messi my place is tirur"

#stores unique words in the text_data
distinct_words = set(text_data.split())
```

```python
def bigram_model(inp):

  #calculates frequency of the input word
  freq_inp = text_data.count(inp)

  if freq_inp==0:
    print("word not found in text data")
    return

  #possible outputs
  outs=[]

  for word in distinct_words:
    comb = inp+" "+word
    freq = text_data.count(comb)

    if freq!=0:
      outs.append([comb,freq/freq_inp])

  if len(outs)==0:
    print(f"{inp} appeared last at the text data,so no following words can be determined")
    return


  print("possible follower word with probability:\n")
  for pair in outs:
    print(str(pair[0])+":"+str(pair[1])+"\n")

  return
```

```python
inp = input("enter the word:")

bigram_model(inp)
```
